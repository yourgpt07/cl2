```python
import pandas as pd

from ucimlrepo import fetch_ucirepo
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Step 1: Load the SMS Spam Collection dataset from UCI
dataset = fetch_ucirepo(id=228)

messages = dataset.data.features.iloc[:, 0].astype(str)
labels = dataset.data.targets.iloc[:, 0].astype(str).str.lower()

# Step 2: Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    messages,
    labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)

# Step 3: Convert text into numerical TF-IDF features
vectorizer = TfidfVectorizer(stop_words="english")

X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)

# Step 4: Train the Naive Bayes classification model
model = MultinomialNB()
model.fit(X_train, y_train)

# Step 5: Test the model and calculate accuracy
y_pred = model.predict(X_test)

print("Model Accuracy:", accuracy_score(y_test, y_pred) * 100, "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Step 6: Predict whether a new message is spam or ham
message = input("\nEnter a message to check: ")

# Convert the input message into TF-IDF features
message_vector = vectorizer.transform([message])

# Predict the message category
prediction = model.predict(message_vector)

print("\nMessage Classification:")

if prediction[0] == "spam":
    print("This message is SPAM.")
else:
    print("This message is HAM (Not Spam).")
```

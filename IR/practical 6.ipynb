```python
import networkx as nx
import matplotlib.pyplot as plt

# Step 1: Create a directed graph
G = nx.DiGraph()

# Step 2: Add web pages and hyperlinks
# For example, page 1 links to pages 2 and 3.
edges = [
    (1, 2),
    (1, 3),
    (2, 3),
    (3, 1),
    (3, 4),
    (4, 3),
    (4, 5),
    (5, 1)
]

G.add_edges_from(edges)

# Step 3: Calculate PageRank using the Random Walk method
# Damping factor = 0.85
pagerank = nx.pagerank(G, alpha=0.85)

# Step 4: Display PageRank scores
print("PageRank Scores:")

for page, score in sorted(pagerank.items()):
    print("Page", page, ":", round(score, 4))

# Step 5: Rank pages according to importance
print("\nWeb Pages Ranked by Importance:")

ranked_pages = sorted(
    pagerank.items(),
    key=lambda x: x[1],
    reverse=True
)

for rank, (page, score) in enumerate(ranked_pages, start=1):
    print(rank, ". Page", page, " - Score:", round(score, 4))

# Step 6: Draw the directed graph
pos = nx.spring_layout(G)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="lightblue",
    node_size=1500,
    arrows=True
)

plt.title("Web Graph using PageRank Algorithm")
plt.show()
```
